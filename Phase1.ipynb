### List vs Numpy array


```python
py_list = [1,2,3]

print("Python List multiplication ", py_list*2)

np_array = np.array([1,2,3]) # element wise multiplication
print("Python array multiplication ", np_array * 2)

import time 
start = time.time()
py_list = [i*2 for i in range (1000000)]
print("\n List operation time: ", time.time() - start)

start = time.time()
np_array = np.arange(1000000) * 2

print("\n Numpy operation time: ", time.time() - start)
```

### Vector, Matriz and Tensor

```python
vector = np.array([1,2,3])
print("Vector array \n", vector)

matrix = np.array([[1,2,3],[4,5,6]])
print("Matrix array \n", matrix)

tensor = np.array([[[1,2], [3,4]],
                   [[5,6], [7,8]]])
print("Tensor array \n", tensor)
```

### Array properties

```python
arr = np.array([[1,2,3],[4,5,6]])
print("Shape of array:" , arr.shape)
print("Dimensions of array:" , arr.ndim)
print("Data type of array:" , arr.dtype)
print("Size of array:" , arr.size)
```

### Array Reshaping 

```python



arr = np.arange(12)
print("Original array: \n", arr)

reshaped = arr.reshape((3,4))
print("Reshaped array: \n", reshaped)

flattened = reshaped.flatten()
print("Flattened array: \n", flattened)
# Ravel returns a flattened array, but it is a view of the original array whenever possible. This means that if you modify the raveled array, it may affect the original array.
raveled = reshaped.ravel()
print("Raveled array: \n", raveled)
```
