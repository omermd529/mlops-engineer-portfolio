```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, roc_auc_score, classification_report
import numpy as np
# Simulate actual fraud labels (1 for fraud, 0 for not-fraud)
# 97% not fraud, 3% fraud realistic distribution

```

```python
np.random.seed(42) 
actual = np.random.choice([0,1], size=1000, p=[0.97,0.03])
# Simulate a model taht catches 80 % of fraud but has some false alarms.
predicted_proba = np.where(actual == 1, 
                           np.random.uniform(0.6,1.0,1000),
                           np.random.uniform(0.0,0.4,1000)
                           )
predicted = (predicted_proba > 0.5 ).astype(int)
print("==== Evaluation Report ====")
print(f"Accuracy: {accuracy_score(actual, predicted):.4f}")
print(f"Precision: {precision_score(actual,predicted):.4f}")
print(f"Recall: {recall_score(actual,predicted):.4f}")
print(f"ROC AUC: {roc_auc_score(actual,predicted):.4f}")
print()
print(classification_report(actual,predicted, target_names=['Not Fraud','Fraud']))
```
